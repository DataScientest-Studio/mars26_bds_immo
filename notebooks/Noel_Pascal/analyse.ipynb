Recuperation du dataset et affichage des informations sur les données

```python
import pandas as pd
import numpy as np
import seaborn as sns

df = pd.read_csv("../../data/raw/dvf.csv")

print(f"\n information {df.info()}")

print(f"\n information description {df.describe()}")

print(f"\n information sur les types de données {df.dtypes}")

df.head(10)
```

Affichage des variables ayant valeur manquante

```python
print(f"\n les variables ayant les valeurs manquantes "+ f"\n {df.isna().sum()/len(df)}")
```

Suppression des variables numeriques ayant plus de 80 pourcents de valeurs manquantes

```python
Val_manquantes = df.isna().sum()/len(df)

cols_valManque = Val_manquantes[Val_manquantes > 0.8].index.to_list()

df = df.drop(columns=cols_valManque)

df.isna().sum()/len(df)
```

Conversion de variable numerique en categorielle

```python
#df['Prix_M2'] = df['valeur_fonciere']/df['surface_reelle_bati']
df['mois'] = pd.to_datetime(df['date_mutation']).dt.month
df['year'] = pd.to_datetime(df['date_mutation']).dt.year
df.dtypes
```

Nettoyage des données 

Partie 1 Identification des données categorielles et numeriques

```python
numerical_cols = df.select_dtypes(include=['int32', 'Int64', 'float64']).columns.tolist()
categorical_cols = df.select_dtypes(include=['object', 'category', 'str']).columns.tolist()

print("📊 Variables numériques:")
print(f"   Nombre: {len(numerical_cols)}")
print(f"   {numerical_cols}\n")

print("🏷️ Variables catégoriques:")
print(f"   Nombre: {len(categorical_cols)}")
print(f"   {categorical_cols}")

df.isna().sum()/len(df)
```

Traitement des valeurs manquantes de la variables valeur_fonciere

```python
# Proportions des valeurs dans la variable valeur_fonciere
print()
print("Proportion des valeurs manquantes: ", end="\n\n")
print(df["valeur_fonciere"].value_counts(dropna=False, normalize=True), end="\n\n")

# Repartition des valeurs manquantes de valeur_fonciere par numero_disposition

df_numerical = df[numerical_cols]

grouped = df_numerical.groupby('numero_disposition').apply(
    lambda x: pd.Series({
        'Total': len(x),
        'NaN': x['valeur_fonciere'].isna().sum(),
        'Pas NaN': x['valeur_fonciere'].notna().sum(),
        'Pourcentage NaN (%)': (x['valeur_fonciere'].isna().sum() / len(x)) * 100
    })
).round(2)
grouped = grouped.sort_index()
print(grouped)
print()

# Le pourcentage de valeur manquante de la variable fonciere est inferieure à 1 pourcent donc on peut aisement supprimer les lignes 

df_numerical = df_numerical.dropna(subset='valeur_fonciere')

#Calcule du mode de valeur_fonciere par No Disposition

#mode_by_disposition = df_numerical.groupby('numero_disposition')['valeur_fonciere'].transform('median')
#df_numerical['valeur_fonciere'] = df_numerical['valeur_fonciere'].fillna(mode_by_disposition)

df_numerical.isna().sum()/len(df_numerical)

sns.boxplot(x='valeur_fonciere',data= df_numerical)

#df[numerical_cols].isna().sum()/len(df)
```

Identification de la correlation des elements 

```python
import matplotlib.pyplot as plt

# Suppression des lignes de valeur manquante
df_numerical = df_numerical.dropna(subset='surface_reelle_bati')



#Calcul de la matrice de correlation

corr = df_numerical.corr()

#HeadMap

fig,ax = plt.subplots(figsize=(12, 8), facecolor= "#0d0d14")
ax.set_facecolor("#1a1a2e")
cmap = sns.diverging_palette(0,200,s=80,l=50,as_cmap=True)

sns.heatmap(
    corr,
    ax=ax,
    annot=True,
    fmt=".2f",
    cmap=cmap,
    center=0,
    vmin=-1, vmax=1,
    linewidths=2,
    linecolor="#0d0d14",
    annot_kws={"size":12,"weight":"bold","color":"gray"},
    cbar_kws={"shrink":0.8,"label":"Pearson r"},
)

# Mise en forme
ax.set_title(
    "Matrice de corrélation",
    fontsize=16, fontweight="bold",
    color="#e2e8f0", pad=16
)
ax.tick_params(colors="#e2e8f0", labelsize=11)
ax.xaxis.label.set_color("#e2e8f0")
ax.yaxis.label.set_color("#e2e8f0")
plt.xticks(rotation=30, ha="right")
plt.yticks(rotation=0)

# Colorbar
cbar = ax.collections[0].colorbar
cbar.ax.yaxis.set_tick_params(color="#64748b")
cbar.set_label("Pearson r", color="#e2e8f0", fontsize=11)
plt.setp(cbar.ax.yaxis.get_ticklabels(), color="#e2e8f0")

plt.tight_layout()
plt.savefig("heatmap_correlation.png", dpi=150,
            bbox_inches="tight", facecolor="#0d0d14")
print("✅ Graphique sauvegardé : heatmap_correlation.png")
plt.show()
```
