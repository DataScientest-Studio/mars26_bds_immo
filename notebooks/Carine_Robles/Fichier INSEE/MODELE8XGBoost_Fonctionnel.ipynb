## Récupération Modèle XGBOOST 1

```python
import xgboost as xgb
import json
import pandas as pd

# 1. Charger la configuration (l'ordre des colonnes, etc.)
with open('config_modele.json', 'r', encoding='utf-8') as f:
    config = json.load(f)

# 2. Initialiser le modèle et charger le fichier JSON
model_recharge = xgb.XGBRegressor()
model_recharge.load_model('modele_immobilier_final.json')

# 3. Récupérer l'ordre exact des colonnes
features_entrainement = config["features_order"]

print("✅ Modèle et configuration chargés avec succès !")
print(f"Le modèle attend {len(features_entrainement)} variables dans cet ordre.")
```

```python
def predire_prix(donnees_entree):
    # Transformation en DataFrame
    df_temp = pd.DataFrame([donnees_entree])
    
    # On s'assure que toutes les colonnes attendues existent (sinon on met 0)
    for col in features_entrainement:
        if col not in df_temp.columns:
            df_temp[col] = 0
            
    # On remet les colonnes dans l'ordre EXACT
    df_temp = df_temp[features_entrainement]
    
    # On prédit
    return model_recharge.predict(df_temp)[0]
```

```python
# Données de la maison en Vendée
maison_vendee = {
    'surface_reelle_bati': 80,
    'nombre_pieces_principales': 4, # Moyenne pour 80m2
    'surface_terrain': 500,        # 5 ares = 500m2
    'latitude': 46.483,
    'longitude': -1.112,
    'is_maison': 1,
    'is_neuf': 0,
    'revenu_median': 21800,
    'population_commune': 2100,
    'dist_hopital': 4.0,
    'dist_lycee': 4.5,
    'dist_supermarche': 3.5,
    'dist_primaire': 1.0,
    'dist_college': 4.5,
    'dist_com_prox': 1.2,
    'dist_sante_prox': 1.5
}

# Utilisation de ta fonction de prédiction
prix_estime_vendee = predire_prix(maison_vendee)

print(f"--- 🏠 TEST MODÈLE : SAINTE-GEMME-LA-PLAINE (85) ---")
print(f"Caractéristiques : 80m², 500m² de terrain")
print(f"Prix estimé par l'IA : {prix_estime_vendee:,.2f} €")
```
