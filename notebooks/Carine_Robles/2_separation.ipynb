```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split

# Chargement du dataset
df = pd.read_csv('dvf_clean_model_ready.csv')

# Séparation des features (X) et de la cible (y)
X = df.drop(columns=['valeur_fonciere'])
y = df['valeur_fonciere']
```

```python
# On garde 20% des données pour le test final
X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.20, 
    random_state=42, 
    shuffle=True
)

print(f"Jeu d'entraînement : {X_train.shape[0]} lignes")
print(f"Jeu de test : {X_test.shape[0]} lignes")
```

```python
# 1. Sauvegarde des Features (X)
X_train.to_csv('X_train.csv', index=False)
X_test.to_csv('X_test.csv', index=False)

# 2. Sauvegarde des Cibles (y) 
# Note : comme y est une Series, on peut aussi l'exporter en CSV
y_train.to_csv('y_train.csv', index=False)
y_test.to_csv('y_test.csv', index=False)

print("Les 4 fichiers de modélisation ont été exportés avec succès.")
```
