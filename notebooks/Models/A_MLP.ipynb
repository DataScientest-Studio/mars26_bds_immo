```python
import pandas as pd
import numpy as np
import time
import matplotlib.pyplot as plt
import seaborn as sns

# Metrics
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
# Model
from sklearn.ensemble import RandomForestRegressor

# Option pour afficher toutes les colonnes
pd.set_option('display.max_columns', None)
```

## Récupération des données

```python
print("Chargement des données en cours...")
start_time = time.time()

X_train = pd.read_csv('X_train.csv')
X_test = pd.read_csv('X_test.csv')

# Pour y, on s'assure d'avoir un vecteur plat (1D)
y_train = pd.read_csv('y_train.csv').values.ravel()
y_test = pd.read_csv('y_test.csv').values.ravel()

load_time = time.time() - start_time
print(f"Données chargées en {load_time:.2f} secondes.")
print(f"Dimensions Train : {X_train.shape} | Dimensions Test : {X_test.shape}")
```

## Stratégie d'entraînement
StandardScaler (comme pour Ridge).

Temps de calcul : Cela peut être long. On va commencer par une architecture raisonnable.

```python
from sklearn.neural_network import MLPRegressor
from sklearn.preprocessing import StandardScaler
import time

# 1. Normalisation (Étape vitale pour les réseaux de neurones)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 2. Configuration du MLP
# On crée 3 couches cachées (100 neurones, 50 neurones, 25 neurones)
mlp = MLPRegressor(
    hidden_layer_sizes=(100, 50, 25), 
    activation='relu', 
    solver='adam', 
    max_iter=50,       # On commence doucement pour tester
    batch_size=1024,   # Pour traiter les 4.6M de lignes plus vite
    random_state=42,
    verbose=True
)

print("Début de l'entraînement du Réseau de Neurones...")
start_time = time.time()
mlp.fit(X_train_scaled, y_train)
print(f"Entraînement terminé en {(time.time() - start_time)/60:.2f} minutes.")

# 3. Évaluation
y_pred_mlp = mlp.predict(X_test_scaled)
r2_mlp = r2_score(y_test, y_pred_mlp)
mae_mlp = mean_absolute_error(y_test, y_pred_mlp)

print(f"--- RÉSULTATS RÉSEAU DE NEURONES ---")
print(f"R² Score : {r2_mlp:.4f}")
print(f"MAE      : {mae_mlp:.2f} €")
```
